```python
from cmath import sqrt
from math import log as ln
from math import exp


S_0 = 80
r = 0.1
K = 100
T = 1/4
delta = 1.5
# black-scholes formula
d_1 = (ln(S_0/K) + (r + delta**2/2)*T)/(delta*sqrt(T))
```

```python

d_2 = d_1 - delta*sqrt(T)
```

```python
# dystruanta N(0,1) - dodaj odpowiednia funkcje/biblioteke
from scipy.stats import norm

# Calculate N(d1) and N(d2) using the cumulative distribution function
N_d1 = norm.cdf(d_1)
N_d2 = norm.cdf(d_2)
```

```python
# Calculate Call Option Price (assuming Black-Scholes formula for Call)
C_0 = S_0 * N_d1 - K * exp(-r * T) * N_d2

print(f"N(d1): {N_d1}")
print(f"N(d2): {N_d2}")
print(f"Call Price: {C_0}")
```

```python
P_0 = K * exp(-r * T) * norm.cdf(-d_2) - S_0 * norm.cdf(-d_1)
```

```python
print(f"Put Price: {P_0}")
```
