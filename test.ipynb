```python
# // json string  load from file response_1.json
with open('response_1.json', 'r') as file:
    json_tags = file.read()
# count tags
import json
data = json.loads(json_tags)
tags = data['tags']
print(f"Total tags: {len(tags)}")

# # check if they are unique
unique_ids = set()

for tag in tags:
    unique_ids.add(tag['id'])

print(f"Unique tags: {len(unique_ids)}")
```
